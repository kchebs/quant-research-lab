# 10 — ABIDES market simulation

Agent-based market simulation with the custom EMA-momentum + inventory-skew agent.
The discrete-event kernel lives under `third_party/abides/` (optional install: `pip install -e ".[abides]"`).


## Agent design

On each wake the agent cancels resting orders, reads the bid/ask, updates fast/slow mid-price EMAs, and only quotes when momentum and inventory logic agree. Near the close it flattens.


```python
from quantlab.abides_runner import abides_available, agent_description, agent_module_path, ensure_abides_on_path
from quantlab.abides_agent import momentum_signal, inventory_skewed_prices

print(agent_description())
print('ABIDES vendored:', abides_available())
print('Agent module:', agent_module_path())
print('Signal on rising mids:', momentum_signal(list(range(1, 40)), 5, 20))
print('Long inventory skew:', inventory_skewed_prices(100, 2, 10_000, 50, 101, 99))
```
