# 12 — Martingale and grid-world Q-learning

Probability experiment (American roulette martingale) and maze navigation with the tabular Q-learner.


```python
from quantlab.sim import run_martingale_experiment, summarize_winnings
import matplotlib.pyplot as plt

uncapped = run_martingale_experiment(n_sims=200, bankroll_cap=None, seed=42)
capped = run_martingale_experiment(n_sims=200, bankroll_cap=256.0, seed=42)
print('Unlimited:', summarize_winnings(uncapped))
print('Capped $256:', summarize_winnings(capped))
```

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from quantlab.rl import GridWorld, QLearner, train_on_grid, greedy_success

world = GridWorld.from_csv(Path('..') / 'data' / 'gridworlds' / 'world01.csv')
learner = QLearner(num_states=world.n_states, num_actions=4, alpha=0.2, gamma=0.9, epsilon=0.5, epsilon_decay=0.999, dyna=50, seed=0)
rewards = train_on_grid(world, learner, episodes=400, max_steps=500)
print('Greedy success on world01:', greedy_success(world, learner))
plt.plot(pd.Series(rewards).rolling(20).mean()); plt.title('Episode reward (rolling)'); plt.show()
```
